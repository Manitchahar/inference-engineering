```python
import sys
for s in (sys.stdout, sys.stderr):
    try:
        s.fileno()
    except Exception:
        s.fileno = lambda: 1

import time
from vllm import LLM, SamplingParams

# Initializing Qwen3-4B
llm = LLM(
    model="Qwen/Qwen3-4B",
    trust_remote_code=True,
    dtype="float16",
    gpu_memory_utilization=0.85,
    max_model_len=2048,
    enforce_eager=True
)

t0 = time.time()
out = llm.generate(
    ["Explain gravity like I'm 5."],
    SamplingParams(max_tokens=60, temperature=0.7)
)
print(f"{time.time()-t0:.1f}s")
print(out[0].outputs[0].text)
```

```python
sampling_params = SamplingParams(
    max_tokens=512,       # Give enough room for thinking + full explanation
    temperature=0.7
)

out = llm.generate(["Explain gravity like I'm 5."], sampling_params)
print(out[0].outputs[0].text)
```

```python
import time
sp = SamplingParams(max_tokens=1, temperature=0.0)
sys = "You are a patient physics tutor for 5-year-olds. Use only tiny words. " * 35
cold = [sys + f"Question {i}: Explain memory bandwidth." for i in range(8)]
warm = [sys + f"Question {i}: Explain compute bandwidth." for i in range(8)]
t0 = time.perf_counter(); llm.generate(cold, sp); t_cold = time.perf_counter() - t0
t0 = time.perf_counter(); llm.generate(warm, sp); t_warm = time.perf_counter() - t0
print(f"COLD: {t_cold:.2f}s | WARM: {t_warm:.2f}s | speedup: {t_cold/t_warm:.2f}x")
```
